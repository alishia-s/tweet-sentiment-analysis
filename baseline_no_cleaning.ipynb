```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import col
```

```python
spark = SparkSession.builder.getOrCreate()
```

```python
train_df = spark.read.csv('data/twitter_training.csv')
```

```python
col_names = ["id", "topic", "sentiment", "content"]
train_df = train_df.toDF(*col_names)
train_df.show(5, truncate=False)
```

```python
raw_df = train_df.withColumn("raw_content", col("content"))
```

```python
raw_df.select("content", "raw_content").show(5, truncate=False)
```

```python
raw_train_df = raw_df.select("id", "topic", "sentiment", "content", "raw_content")
raw_train_df.show(5, truncate=False)
```
